```python
#creat a histogram of the 'fixed acidity' feature

plt.figure(figsize=(8,6))
sns.histplot(data = data,x='fixed acidity',color='yellow',edgecolor='linen',
                  alpha=0.5,bins=5)

plt.title('fixed acidity feature')
plt.xlabel('fixed acidity')
plt.ylabel('count')
plt.show()
```

```python
#calculate the mean() in 'fixed acidity' feature

round(data['fixed acidity'].mean())
```

```python
#calculate the median() in'fixed acidity'feature

print(data['fixed acidity'].median())
```

```python
#hitogramplot of 'fixed acidity' feature to show mean()and median()

plt.figure(figsize=(8,6))

sns.histplot(data = data,x='fixed acidity',color='yellow',edgecolor='linen',
             alpha=0.5,bins=5)

plt.title('histogram fixed acidity')
plt.xlabel('fixed acidity')
plt.ylabel('count')
plt.vlines(data['fixed acidity'].mean(),ymin=0,ymax=4500,color='blue',label='mean')
plt.vlines(data['fixed acidity'].median(),ymin=0,ymax=4500,color='green',label='median')
plt.legend()
plt.show()
```

```python
#histogramplot 'volatile acidity' feature

plt.figure(figsize=(8,6))

sns.histplot(data=data,x='volatile acidity',color='blue',edgecolor='red',alpha=0.5,
            bins=5)
plt.title('volatile acidity feature')
plt.xlabel('volatile acidity')
plt.ylabel('count')
plt.show()
```

```python
# displot 'volatile acidity' feature

plt.figure(figsize=(11,6))

sns.displot(data['volatile acidity'],color='blue')
plt.title('displot volatile acitity feature')
plt.xlabel('volatile Acitity')
plt.ylabel('density')
plt.show()
```

```python
# calculate swekness of 'volatile acidity'
print(data['volatile acidity'].skew())
```

```python
#calculate the mean of 'volatile acidity'feature
print(data['volatile acidity'].mean())
```

```python
# histogram of the 'volatile acidity' to show a mean ana medain

plt.figure(figsize=(8,6))

sns.histplot(data=data,x='volatile acidity',color='blue',edgecolor='red',alpha=0.5,
            bins=5)
plt.title('histogram volatile acidity feature')
plt.xlabel('volatile acidity')
plt.ylabel('destity')

plt.vlines(data['volatile acidity'].mean(),ymin=0,ymax = 4000,color = 'pink',label = 'mean')
plt.vlines(data['volatile acidity'].median(),ymin=0,ymax = 4000,color = 'yellow',label = 'median')
plt.legend()
plt.show()
```

```python
# histogramplot of the 'citric acid' feature

plt.figure(figsize=(8,6))
sns.histplot(data=data,x='citric acid',color = 'black',edgecolor = 'yellow',alpha=0.5,
            bins= 5)
plt.title('histogram citric acid')
plt.xlabel('citric acid')
plt.ylabel('count')
plt.show()
```

```python
#calculate the mean to 'citric acid'feature
print(data['citric acid'].mean())
```

```python
#calculate the median'citric acid'feature
print(data['citric acid'].median())
```

```python
#histogram of the'citric acid' feature to show mean and median
plt.figure(figsize=(8,6))
sns.histplot(data=data,x='citric acid',color='black',edgecolor='yellow',alpha=0.5,
             bins=5)
plt.title('histogram the citric acid')
plt.xlabel('citric acid')
plt.ylabel('count')

plt.vlines(data['citric acid'].mean(),ymin=0,ymax=4000,color = 'orange',label='mean')
plt.vlines(data['citric acid'].median(),ymin=0,ymax=4000,color = 'skyblue',label='median')
plt.legend()
plt.show()
```

```python
# calculate displot using'citric acidity'feature
plt.figure(figsize=(8,6))

sns.displot(data['citric acid'],color='black')
plt.title('distplot citric acid')
plt.xlabel('citric acid')
plt.ylabel('count')
plt.show()
```

```python
#countplot of the 'quality' feature
plt.figure(figsize=(11,6))

sns.countplot(data=data,x='quality')

plt.title('countplot of quality')
plt.xlabel('quality')
plt.ylabel('count')
plt.show()
```
